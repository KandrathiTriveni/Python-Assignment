```python
start_reading = 11236
end_reading = 11494
per_km_charge = 15
tollgate_charge = 200
total_travelled_km = end_reading - start_reading
travel_amt = total_travelled_km * per_km_charge
final_bill = travel_amt + tollgate_charge
print('Total travelled km ', total_travelled_km)
print('Travel amount ', travel_amt)
print('Final bill ',final_bill)
```
